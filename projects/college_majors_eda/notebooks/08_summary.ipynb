## Итоговая интерпретация


### Зарплатные результаты

**Категория специальности — главный предиктор зарплаты**    
Медианные зарплаты значимо различаются между категориями. Наибольшие — у `Engineering`, наименьшие — у `Education`.

**STEM зарабатывает значимо больше Non-STEM**   
Различие статистически и практически значимо. Разрыв обусловлен концентрацией высокооплачиваемых направлений в Engineering, Computers & Mathematics и Physical Sciences.

**Медианная зарплата и занятость полный день тесно связаны**    
Наиболее сильная корреляция из проверенных. Чем выше медианная зарплата направления, тем выше доля работников на полной ставке — обе характеристики отражают «качество» специальности на рынке труда.

**Разброс зарплат внутри категорий существенно различается**    
Наиболее предсказуемы зарплаты в `Communications & Journalism` и `Arts`; наименее — в `Engineering` и `Health`.


### Уровни образования

**Аспирантура даёт существенную прибавку к зарплате**   
Медианная зарплата аспирантов выше, чем у выпускников без неё. Преимущество устойчиво во всех специальностях без исключения.

**Аспирантура значимо снижает безработицу**        
Безработица аспирантов ниже, чем у выпускников без степени. Эффект наблюдается независимо от категории направления.

**Зарплатная премия за аспирантуру неодинакова по категориям**  
Наибольшая финансовая выгода — у `Biology & Life Science`, а наименьшая — у `Engineering`, поскольку инженеры и без степени зарабатывают высоко.

**Потребность в продолжении обучения кардинально различается между категориями**    
В `Biology & Life Science` бакалавриат — лишь первый шаг; в `Industrial Arts & Consumer Services` — финальная ступень.

**Аспиранты чаще работают полный рабочий день круглый год**     
Высшая степень обеспечивает бо́льшую стабильность занятости.


### Гендерный анализ

**Специальности с преобладанием мужчин зарабатывают значимо больше**    
Медиана у направлений с `ShareWomen` < 0.3 выше, чем у направлений с `ShareWomen` > 0.7.

**Доля женщин сильно отрицательно коррелирует с зарплатой**     
Чем выше доля женщин в направлении, тем ниже медианная зарплата. Связь устойчивее, чем влияние STEM-статуса.

**Категория специальности сильно определяет долю женщин**   
Распределение `ShareWomen` по категориям статистически высоко значимо: гендерная сегрегация по специальностям выражена и системна.

**Доля женщин умеренно отрицательно связана с занятостью полный день**      
Направления с высокой долей женщин характеризуются меньшей долей работников на полной ставке.
