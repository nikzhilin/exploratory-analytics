## Итоговая интерпретация


### Зарплатные результаты

**Категория специальности — главный предиктор зарплаты** (η² = 0.66, большой эффект).  
Медианные зарплаты значимо различаются между категориями (p ≈ 0). Наибольшие — у `Engineering`, наименьшие — у `Education`.

**STEM зарабатывает значимо больше Non-STEM** (большой эффект по Cohen).  
Различие статистически и практически значимо. Разрыв обусловлен концентрацией высокооплачиваемых направлений в Engineering, Computers & Mathematics и Physical Sciences.

**Медианная зарплата и занятость полный день тесно связаны** (ρ = 0.76, p ≈ 0, большой эффект).  
Наиболее сильная корреляция из проверенных. Чем выше медианная зарплата направления, тем выше доля работников на полной ставке — обе характеристики отражают «качество» специальности на рынке труда.

**Разброс зарплат внутри категорий существенно различается** (тест Левена p = 0.019, η² = 0.15).  
Наиболее предсказуемы зарплаты в `Communications & Journalism` и `Arts`; наименее — в `Engineering` и `Health`.

Отвергнутые гипотезы: массовость специальности не связана с медианным доходом (p = 0.089).

### Уровни образования

**Аспирантура даёт существенную прибавку к зарплате** (r = 0.85, очень большой эффект).  
Медианная зарплата аспирантов — $75 000 против $55 000 у выпускников без неё (p ≈ 0). Преимущество устойчиво во всех специальностях без исключения.

**Аспирантура значимо снижает безработицу** (r = 0.63, большой эффект).  
3.95% против 5.38% у выпускников без продвинутой степени (p ≈ 0). Эффект наблюдается независимо от категории направления.

**Зарплатная премия за аспирантуру неодинакова по категориям** (η² = 0.33).  
Наибольшая финансовая выгода — у `Biology & Life Science` (медианная премия ≈ 52%); наименьшая — у `Engineering` (≈ 21%), поскольку инженеры и без степени зарабатывают высоко.

**Потребность в продолжении обучения кардинально различается между категориями** (Краскел–Уоллис p ≈ 0, η² = 0.45).  
В `Biology & Life Science` бакалавриат — лишь первый шаг; в `Industrial Arts & Consumer Services` — финальная ступень.

**Аспиранты чаще работают полный рабочий день круглый год** (r = 0.31, средний эффект).  
Высшая степень обеспечивает бо́льшую стабильность занятости.

Отвергнутая гипотеза: PhD не снижает относительный разброс доходов (p = 0.239, r = 0.05 — тривиальный эффект).

### Гендерный анализ

**Специальности с преобладанием мужчин зарабатывают значимо больше** (r = −0.89, очень большой эффект).  
Медиана у направлений с `ShareWomen` < 0.3 — $50 000; у направлений с `ShareWomen` > 0.7 — $33 000 (p ≈ 0). Это наибольший размер эффекта во всём анализе.

**Доля женщин сильно отрицательно коррелирует с зарплатой** (ρ = −0.66, p ≈ 0, большой эффект).  
Чем выше доля женщин в направлении, тем ниже медианная зарплата. Связь устойчивее, чем влияние STEM-статуса.

**Категория специальности сильно определяет долю женщин** (η² = 0.62, большой эффект).  
Распределение `ShareWomen` по категориям статистически высоко значимо (p ≈ 0): гендерная сегрегация по специальностям выражена и системна.

**Доля женщин умеренно отрицательно связана с занятостью полный день** (ρ = −0.40, p ≈ 0, средний эффект).  
Направления с высокой долей женщин характеризуются меньшей долей работников на полной ставке.

Отвергнутые гипотезы: доля женщин не связана с уровнем безработицы (p = 0.390) и долей трудоустроенных по специальности (p > 0.05).
