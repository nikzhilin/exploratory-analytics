# Итоговые выводы


Цель данного ноутбука - собрать в одном месте ключевые результаты предыдущих этапов и зафиксировать общую интерпретацию структуры рынка Airbnb в Ашвилле.


## Итоговая интерпретация


### Рыночная структура и география
Ценовой слой рынка определяется сразу несколькими факторами. Масштаб хоста положительно связан с ценой, но лишь умеренно, тогда как `minimum_nights` почти не объясняет ценовые различия. Крупные игроки реже скрывают цену, чаще попадают в верхний ценовой хвост и поддерживают более высокую доступность объектов. На географическом уровне рынок неоднороден: зона `28801` сильнее насыщена индустриальным сегментом, а районы `28805` и `28806` заметно больше опираются на индивидуальных хостов.


### Хосты и спрос
Большинство владельцев на рынке остаются индивидуальными, однако существенная доля предложения сосредоточена у малого числа крупных хостов. Внутри основных типов жилья крупные игроки удерживают более высокий медианный ценовой уровень и чаще задают более короткий минимальный срок проживания, нежели остальные сегменты. По `review_intensity` крупные и средние хосты выглядят немного сильнее индивидуальных, но сама зависимость текущего спроса от доступности остаётся слабой.


## Финальные выводы
1. Рынок Airbnb в Ашвилле формально раздроблен по числу хостов, но заметная часть предложения и премиального сегмента контролируется относительно небольшим числом крупных игроков.
2. Для объяснения цены важнее сочетание масштаба хоста, географии и рыночного позиционирования, чем требования по минимальному сроку проживания.
3. Производные признаки подтверждают, что рынок неоднороден одновременно по структуре предложения, прозрачности цен и свежести спроса, поэтому дальнейшие выводы следует делать с учётом сегмента хоста и района.
