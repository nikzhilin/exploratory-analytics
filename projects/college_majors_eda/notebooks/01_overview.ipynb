# Обзор данных


Зафиксируем общий состав наборов данных и получим первичное понимание структуры признаков и общего контекста.

Наборы данных представляют собой агрегированную информацию о доходах, занятости, поле и уровне образования выпускников различных направлений колледжей в Америке, собранные агенством American Community Survey.


## Загрузка библиотек

```python
from pathlib import Path

import pandas as pd
from ydata_profiling import ProfileReport
import warnings

warnings.filterwarnings("ignore")
```

## Введение констант

```python
RAW_DATA_DIR = Path("../data/raw/")
PROFILES_DIR = Path("../artifacts/profiles")

DATASET_FILES = {
    "majors_all_ages": RAW_DATA_DIR / "majors_all_ages.csv",
    "majors_degree_levels": RAW_DATA_DIR / "majors_degree_levels.csv",
    "majors_recent_graduates": RAW_DATA_DIR / "majors_recent_graduates.csv",
}
```

## Загрузка наборов данных

```python
datasets = {name: pd.read_csv(path) for name, path in DATASET_FILES.items()}
```

## Анализ набора данных `majors_all_ages`

Набор данных хранит общую информацию о доходах и занятости выпускников

### Схема набора данных

```python
datasets["majors_all_ages"].info()
```

**Информация о программе**  
**Major_code** - код, идентификатор программы согласно FOD1P   
**Major** - название направления обучения   
**Major_category** - название общего направления подготовки 

**Трудоустройство выпускников**  
**Total** - общее количество выпускников  
**Employed** - количество трудоустроенных выпускников   
**Employed_full_time_year_round** - количество выпускников с полной занятостью круглый год    
**Unemployed** - количество нетрудоустроенных выпускников   
**Unemployment_rate** - уровень безработицы  

**Зарплатные характеристики**   
**Median** - медианная зарплата   
**P25th** - зарплата 25% перцентиля   
**P75th** - зарплата 75% перцентиля

### Первичный анализ

```python
profile = ProfileReport(datasets["majors_all_ages"], minimal=True)
profile.to_file(PROFILES_DIR / "majors_all_ages.html")
profile.to_notebook_iframe()
```

### Промежуточные выводы

1. Значения в наборе данных соответствуют логике
2. Отдельного анализа и ограничений требуют выбросы в доходах и занятости

## Анализ набора данных `majors_degree_levels`

Набор данных хранит сведения о доходах и занятости выпускников различного уровня образования

### Схема набора данных

```python
datasets["majors_degree_levels"].info()
```

**Информация о программе**  
**Major_code** - код, идентификатор программы согласно FOD1P   
**Major** - название направления обучения   
**Major_category** - название общего направления подготовки 

**Характеристики выпускников с PhD**  
**Grad_total** - общее количество выпускников с PhD  
**Grad_sample_size** - размер выборки выпускников с PhD  
**Grad_employed** - количество трудоустроенных выпускников с PhD  
**Grad_full_time_year_round** - количество выпускников с полной занятостью круглый год  
**Grad_unemployed** - количество безработных выпускников  
**Grad_unemployment_rate** - уровень безработицы выпускников с PhD  
**Grad_median** - медианная зарплата выпускников с PhD  
**Grad_P25** - зарплата 25% перцентиля выпускников с PhD  
**Grad_P75** - зарплата 75% перцентиля выпускников с PhD 

**Характеристики выпускников без PhD**  
**Nongrad_total** - общее количество выпускников без PhD  
**Nongrad_employed** - количество трудоустроенных выпускников без PhD  
**Nongrad_full_time_year_round** - количество выпускников с полной занятостью круглый год без PhD  
**Nongrad_unemployed** - количество безработных выпускников без PhD  
**Nongrad_unemployment_rate** - уровень безработицы выпускников без PhD  
**Nongrad_median** - медианная зарплата выпускников без PhD  
**Nongrad_P25** - зарплата 25% перцентиля выпускников без PhD  
**Nongrad_P75** - зарплата 75% перцентиля выпускников без PhD 

**Сравнительные показатели**  
**Grad_share** - доля выпускников с PhD  
**Grad_premium** - премия к зарплате за PhD

### Первичный анализ

```python
profile = ProfileReport(datasets["majors_degree_levels"], minimal=True)
profile.to_file(PROFILES_DIR / "majors_degree_levels.html")
profile.to_notebook_iframe()
```

### Промежуточные выводы

1. Значения в наборе данных соответствуют логиче
2. Отдельного анализа и ограничений требуют выбросы в доходах и занятости   
3. Необычным выглядит наличие отрицательных значений признака `Grad_premium`, необходима отдельная работа с ним

## Анализ набора данных `majors_recent_graduates`

Набор данных хранит сведения о доходах и занятости выпускников различного пола

### Схема набора данных

```python
datasets["majors_recent_graduates"].info()
```

**Информация о программе**  
**Rank** - ранг программы по медианной зарплате   
**Major_code** - код, идентификатор программы согласно FOD1P   
**Major** - название направления обучения   
**Major_category** - название общего направления подготовки 

**Информация о выборке**  
**Total** - общее количество выпускников  
**Men** - количество мужчин  
**Women** - количество женщин  
**ShareWomen** - доля женщин  

**Трудоустройство выпускников**  
**Employed** - количество трудоустроенных выпускников  
**Full_time** - количество выпускников с полной занятостью      
**Part_time** - количество выпускников с неполной занятостью        
**Full_time_year_round** - количество выпускников с полной занятостью круглый год   
**Unemployed** - количество безработных выпускников  
**Unemployment_rate** - уровень безработицы 

**Зарплатные характеристики**  
**Sample_size** - размер выборки людей, для которых подсчитываются зарплатные характеристики    
**Median** - медианная зарплата  
**P25th** - зарплата 25% перцентиля  
**P75th** - зарплата 75% перцентиля 

**Тип занятости**  
**College_jobs** - работы, требующие образования      
**Non_college_jobs** - работы, не требующие образования  
**Low_wage_jobs** - низкооплачиваемые работы

### Первичный анализ

```python
profile = ProfileReport(datasets["majors_recent_graduates"], minimal=True)
profile.to_file(PROFILES_DIR / "majors_recent_graduates.html")
profile.to_notebook_iframe()
```

### Промежуточные выводы

1. Значения в наборе данных соответствуют логиче
2. Отдельного анализа и ограничений требуют выбросы в доходах и занятости   
3. Многие признаки имеют нулевые значения, которые не являются типичными для большинства значений
4. Неизвестное количество студентов одного из направлений требует отдельного анализа
