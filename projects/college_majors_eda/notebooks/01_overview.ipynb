### Анализ признаков

### Дубликаты


```python
for dataset_name, dataset_object in datasets.items():
    print(f"Dataset: {dataset_name}, number of duplicates: {dataset_object.duplicated().sum()}")
```

### Пропущенные значения

```python
for dataset_name, dataset_object in datasets.items():
    print(f"Dataset: {dataset_name}, any missings: {dataset_object.isna().sum().any()}")

```

```python
for dataset_name, dataset_object in datasets.items():
    if dataset_object.isna().sum().any() == True:
        print(f"Dataset: {dataset_name}\nRows with nulls:\n{dataset_object[dataset_object.isna().any(axis=1)]}")
        print()
    
```

### Распределение медианных зарплат

```python
sns.set_theme("paper")

MEDIAN_PER_DATASET = {
    "all_ages" : "Median",
    "grad_students" : "Grad_median",
    "recent_grads" : "Median"
}


for dataset_name, median_name in MEDIAN_PER_DATASET.items():
    dataset_object = datasets[dataset_name]
    sns.displot(data=dataset_object, x=median_name, bins=30, kde=True)

```

```python
sns.set_theme("paper")

MEDIAN_PER_DATASET = {
    "all_ages" : "Median",
    "grad_students" : "Grad_median",
    "recent_grads" : "Median"
}


for dataset_name, median_name in MEDIAN_PER_DATASET.items():
    dataset_object = datasets[dataset_name]
    plt.figure(figsize=(10, 2))
    sns.boxplot(x=dataset_object[median_name])
    plt.show()

```

### Распределение безработицы

```python
UNEMPLOYMENT_PER_DATASET = {
    "all_ages" : "Unemployment_rate",
    "grad_students" : "Grad_unemployment_rate",
    "recent_grads" : "Unemployment_rate"
}

for dataset_name, unemployment_percentage in UNEMPLOYMENT_PER_DATASET.items():
    dataset_object = datasets[dataset_name]
    sns.displot(data=dataset_object, x=unemployment_percentage, bins=30, kde=True)
```

```python
for dataset_name, unemployment_percentage in UNEMPLOYMENT_PER_DATASET.items():
    dataset_object = datasets[dataset_name]
    print(f"Dataset: {dataset_name}")
    display(dataset_object[["Major", unemployment_percentage]].sort_values(unemployment_percentage, ascending=False).head(10))
    
```

```python
for dataset_name, unemployment_percentage in UNEMPLOYMENT_PER_DATASET.items():
    dataset_object = datasets[dataset_name]
    print(f"Dataset: {dataset_name}")
    display(dataset_object[["Major", unemployment_percentage]].sort_values(unemployment_percentage, ascending=True).head(10))
    
```
