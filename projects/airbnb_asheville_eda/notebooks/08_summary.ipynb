# Итоговые выводы


Цель данного ноутбука - собрать в одном месте ключевые результаты предыдущих этапов и зафиксировать общую интерпретацию структуры рынка Airbnb в Ашвилле.


## Итоговая интерпретация


### Рыночная структура и география
На полной выборке линейная связь `price` с масштабом хоста и `availability_ratio` оказывается слабой, а `minimum_nights` почти не влияет на цену. При этом медианные цены и уровень доступности у крупных хостов выше, а доля `price`-выбросов особенно заметна у `Industrial host`. Географически индустриальный сегмент сильнее всего проявляется в `28801`, тогда как максимум ценовых выбросов приходится на `28806` и `28801`.


### Хосты и спрос
Большинство владельцев по-прежнему остаются индивидуальными, но заметная доля предложения сосредоточена у более крупных игроков. Внутри массовых типов жилья крупные хосты чаще работают в более дорогом сегменте, хотя эффект неодинаков по `room_type`. По свежей отзывной активности различия умеренные: опытные и крупные сегменты немного лучше удерживают текущий спрос, а отсутствие цены сильнее связано с форматом размещения, чем с географией.


## Финальные выводы
1. Рынок Airbnb в Ашвилле раздроблен по числу хостов, но существенная часть предложения и верхнего ценового хвоста приходится на более крупных игроков.
2. Для объяснения `price` недостаточно одной линейной связи с размером хоста или доступностью: на полной выборке эти зависимости слабы, хотя медианные цены у крупных сегментов выше.
3. `minimum_nights` почти не связан с ценой, зато `availability_ratio`, `review_intensity` и тип хоста лучше раскрывают различия в структуре предложения.
4. Пространственная карта рынка неоднородна: `28801` сочетает высокую долю индустриальных хостов с повышенными ценами, а `28806` вместе с ним концентрируют значимую часть ценовых выбросов.
